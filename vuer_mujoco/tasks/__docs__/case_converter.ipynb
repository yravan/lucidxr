# Case Converter

This notebook provides functions to convert between different string case formats:
- PascalCase to snake_case
- snake_case to PascalCase
- Additional case conversion utilities

```python
def pascal_to_snake(pascal_str):
    """
    Convert a PascalCase string to snake_case.

    Args:
        pascal_str (str): The PascalCase string to convert

    Returns:
        str: The snake_case version of the input string
    """
    if not pascal_str:
        return ""

    result = [pascal_str[0].lower()]
    for char in pascal_str[1:]:
        if char.isupper():
            result.append('_')
            result.append(char.lower())
        else:
            result.append(char)
    return ''.join(result)
```

```python
def snake_to_pascal(snake_str):
    """
    Convert a snake_case string to PascalCase.

    Args:
        snake_str (str): The snake_case string to convert

    Returns:
        str: The PascalCase version of the input string
    """
    if not snake_str:
        return ""

    components = snake_str.split('_')
    return ''.join(x.title() for x in components)
```

```python
# Test the case conversion functions
test_pascal = "HelloWorld"
test_snake = "hello_world"

print(f"PascalCase to snake_case: {test_pascal} -> {pascal_to_snake(test_pascal)}")
print(f"snake_case to PascalCase: {test_snake} -> {snake_to_pascal(test_snake)}")
```

```python
def camel_to_snake(camel_str):
    """
    Convert a camelCase string to snake_case.

    Args:
        camel_str (str): The camelCase string to convert

    Returns:
        str: The snake_case version of the input string
    """
    if not camel_str:
        return ""

    result = [camel_str[0].lower()]
    for char in camel_str[1:]:
        if char.isupper():
            result.append('_')
            result.append(char.lower())
        else:
            result.append(char)
    return ''.join(result)
```

```python
def snake_to_camel(snake_str):
    """
    Convert a snake_case string to camelCase.

    Args:
        snake_str (str): The snake_case string to convert

    Returns:
        str: The camelCase version of the input string
    """
    if not snake_str:
        return ""

    components = snake_str.split('_')
    return components[0] + ''.join(x.title() for x in components[1:])
```

```python
# Test the additional case conversion functions
test_camel = "helloWorld"

print(f"camelCase to snake_case: {test_camel} -> {camel_to_snake(test_camel)}")
print(f"snake_case to camelCase: {test_snake} -> {snake_to_camel(test_snake)}")
```

## Comprehensive Test Cases

```python
def test_case_conversions():
    # Test cases
    test_cases = [
        {"pascal": "HelloWorld", "snake": "hello_world", "camel": "helloWorld"},
        {"pascal": "User", "snake": "user", "camel": "user"},
        {"pascal": "HTTPRequest", "snake": "http_request", "camel": "httpRequest"},
        {"pascal": "APIController", "snake": "api_controller", "camel": "apiController"},
        {"pascal": "FirstNameLastName", "snake": "first_name_last_name", "camel": "firstNameLastName"}
    ]

    print("Testing PascalCase to snake_case:")
    for case in test_cases:
        result = pascal_to_snake(case["pascal"])
        print(f"  {case['pascal']} -> {result} ({'✓' if result == case['snake'] else '✗'})")

    print("\nTesting snake_case to PascalCase:")
    for case in test_cases:
        result = snake_to_pascal(case["snake"])
        print(f"  {case['snake']} -> {result} ({'✓' if result == case['pascal'] else '✗'})")

    print("\nTesting snake_case to camelCase:")
    for case in test_cases:
        result = snake_to_camel(case["snake"])
        print(f"  {case['snake']} -> {result} ({'✓' if result == case['camel'] else '✗'})")

    print("\nTesting camelCase to snake_case:")
    for case in test_cases:
        result = camel_to_snake(case["camel"])
        print(f"  {case['camel']} -> {result} ({'✓' if result == case['snake'] else '✗'})")

test_case_conversions()
```

```python
def kebab_to_snake(kebab_str):
    """
    Convert a kebab-case string to snake_case.

    Args:
        kebab_str (str): The kebab-case string to convert

    Returns:
        str: The snake_case version of the input string
    """
    if not kebab_str:
        return ""

    return kebab_str.replace('-', '_')
```

```python
def snake_to_kebab(snake_str):
    """
    Convert a snake_case string to kebab-case.

    Args:
        snake_str (str): The snake_case string to convert

    Returns:
        str: The kebab-case version of the input string
    """
    if not snake_str:
        return ""

    return snake_str.replace('_', '-')
```

```python
# Test kebab case conversion
test_kebab = "hello-world"
print(f"kebab-case to snake_case: {test_kebab} -> {kebab_to_snake(test_kebab)}")
print(f"snake_case to kebab-case: {test_snake} -> {snake_to_kebab(test_snake)}")
```
